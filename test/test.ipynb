```python
import sys
import os
import scanpy as sc
# Parent directory of Nutkin
project_root = os.path.abspath("..")  # the parent directory of the test folder
sys.path.append(project_root)

import nutkin.preprocessing
```

```python
data_path="...Nutkin/data/test_data.h5ad"
adata = sc.read_h5ad(data_path)
print(adata)
```

```python
output_path=".../Nutkin/test/result" 
```

```python
adata = nutkin.preprocessing.load_data(
        data_path=data_path,
        output_path=output_path,
        n_comps=50,
        run_pca=True,
        run_embedding=True,
        group_col='clone_id'
)
```

```python
from nutkin.nutkin import Nutkin
```

```python
adata = sc.read(".../Nutkin/test/result/preprocessed_data.h5ad")
```

```python
nut = Nutkin(
        adata=adata,
        group_col='clone_id',
        output_path=output_path,
        detail=True
    )
```

```python
nut.measure()
```

```python
nut.differential_test(group1=['mCHERRY_Barcode_1614','mCHERRY_Barcode_1614'],group2=['mCHERRY_Barcode_5774','mCHERRY_Barcode_1755'])
```

```python
nut.differential_test()
```

```python
nut.visualize_pca_results(detail=True, plot_type='both')
```
